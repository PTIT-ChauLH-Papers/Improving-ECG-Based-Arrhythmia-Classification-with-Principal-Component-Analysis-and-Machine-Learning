### Read csv to extract accuracy of 4 classes

```python
criteria = range(90,100,1)
MLs = ["Ada", "DT", "KNN", "LR", "RF", "SVM", "XGB"]
res = {"DT": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}, "Ada": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}, "KNN": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}, "LR": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}, 
"RF": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}, "SVM": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}, "XGB": {'AFIB': 0, 'SB': 0, 'SR': 0, 'GSVT' : 0}}

for ML in MLs:
    for number in criteria:
        df = pd.read_csv(f"./Result/{ML}_0_{number}.csv")
        res[ML]['AFIB'] += df.iloc[0,2]
        res[ML]['SB'] += df.iloc[1,2]
        res[ML]['SR'] += df.iloc[2,2]
        res[ML]['GSVT'] += df.iloc[3,2]

    print(f"Algo: {ML}")
    print(f"AFIB: {(res[ML]['AFIB'] / len(criteria)) * 100}")
    print(f"SB: {(res[ML]['SB'] / len(criteria)) * 100}")
    print(f"SR: {(res[ML]['SR'] / len(criteria)) * 100}")
    print(f"GSVT: {(res[ML]['GSVT'] / len(criteria)) * 100}" )
    print("====")

```
