```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Dữ liệu mẫu
variance = [90, 95, 96, 98, 99]
model = ['Ada_0_', 'DT_0_', 'KNN_0_', 'LR_0_', 'RF_0_', 'SVM_0_', 'XGB_0_']

# Tạo ma trận dữ liệu để vẽ heatmap
heatmap_data = np.zeros((len(model), len(variance)))

for i, m in enumerate(model):
    for j, v in enumerate(variance):
        namefile = f'./Result/{m}{v}.csv'
        df_res = pd.read_csv(namefile)
        
        # Lấy giá trị micro avg
        heatmap_data[i, j] = df_res.iloc[:3, 2].mean()

# Tạo DataFrame cho dữ liệu heatmap
heatmap_df = pd.DataFrame(heatmap_data, columns=variance, index=model)

# Vẽ heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(heatmap_df, annot=True, cmap='Blues', fmt='.2f', cbar_kws={'label': 'Micro Average'})
plt.title("Biểu đồ Heatmap - Micro Average")
plt.show()
```

```python
import seaborn as sns

# Giả sử bạn đã có dữ liệu
variance = [90, 95, 96, 98, 99]
model = ['Ada_0_', 'DT_0_', 'KNN_0_', 'LR_0_', 'RF_0_', 'SVM_0_', 'XGB_0_']

# Chuẩn bị dữ liệu cho boxplot
boxplot_data = []
model_labels = []

for m in model:
    for v in variance:
        namefile = f'./Result/{m}{v}.csv'
        df_res = pd.read_csv(namefile)
        
        # Lấy dữ liệu micro avg
        micro_avg = df_res.iloc[:3, 2].mean()
        boxplot_data.append(micro_avg)
        model_labels.append(m[:-3])  # Lưu tên mô hình

# Chuyển đổi thành DataFrame
df_boxplot = pd.DataFrame({'Model': model_labels, 'Micro Average': boxplot_data})

# Vẽ boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x="Model", y="Micro Average", data=df_boxplot)
plt.ylim([0.90, 1])
plt.title("Biểu đồ Boxplot - Micro Average")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Giả sử bạn có dữ liệu như trước
variance = [90, 95, 96, 98, 99]
model = ['Ada_0_', 'DT_0_', 'KNN_0_', 'LR_0_', 'RF_0_', 'SVM_0_', 'XGB_0_']

for m in model:
    res_variance = []
    for v in variance:
        namefile = f'./Result/{m}{v}.csv'
        df_res = pd.read_csv(namefile)
        
        # Lấy giá trị micro avg từ cột thứ 3
        res_variance.append(df_res.iloc[:3, 2].mean())

    # Vẽ biểu đồ scatter
    plt.scatter(np.arange(0, 5), res_variance, label=m[:-3])

plt.ylim([0.90, 1])
plt.xticks(np.arange(0, 5), labels=['0.90', '0.95', '0.96', '0.98', '0.99'])
plt.legend(loc="upper right", ncols=4)
plt.title("Biểu đồ Scatter - Micro Average")
plt.xlabel("Variance")
plt.ylabel("Micro Average")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Giả sử bạn có dữ liệu như trên
variance = [90, 95, 96, 98, 99]
model = ['Ada_0_', 'DT_0_', 'KNN_0_', 'LR_0_', 'RF_0_', 'SVM_0_', 'XGB_0_']

# Tạo dữ liệu cho Stacked Bar
data = []

for m in model:
    res_variance = []
    for v in variance:
        namefile = f'./Result/{m}{v}.csv'
        df_res = pd.read_csv(namefile)
        
        # Lấy giá trị micro avg
        res_variance.append(df_res.iloc[:3, 2].mean())
    
    data.append(res_variance)

# Chuyển đổi dữ liệu thành mảng numpy
data = np.array(data)

# Vẽ Stacked Bar Chart với điều chỉnh khoảng cách cột
plt.figure(figsize=(10, 6))
bar_width = 0.4  # Đặt độ rộng của cột
index = np.arange(len(variance))  # Tạo mảng chỉ số cho trục x

# Vẽ cột cho mỗi mô hình với khoảng cách thích hợp
plt.bar(index, data[0], bar_width, label=model[0][:-3])
for i in range(1, len(model)):
    plt.bar(index, data[i], bar_width, bottom=data[:i].sum(axis=0), label=model[i][:-3])

# Đặt nhãn cho trục x
plt.xticks(index, variance)

# Thiết lập nhãn trục
plt.xlabel("Variance")
plt.ylabel("Micro Average")
plt.title("Biểu đồ Stacked Bar - Micro Average")
plt.ylim((0,8))
# Hiển thị legend
plt.legend(loc="upper left", ncol=4)

# Hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

variance = [90, 95, 96, 98, 99, 100]
model = ['Ada_0_', 'DT_0_', 'KNN_0_', 'LR_0_', 'RF_0_', 'SVM_0_', 'XGB_0_']

for m in model:
    res_variance = []
    for v in variance:
        namefile = f'./Result/{m}{v}.csv'
        df_res = pd.read_csv(namefile)
        res_variance.append(df_res.iloc[5, 3])
    
    label_name = m[:-3]
    x = np.arange(len(variance))
    
    # Scatter points
    plt.scatter(x, res_variance, label=label_name)
    # Nối các điểm bằng đường line
    plt.plot(x, res_variance)

plt.ylim([0.5, 1])
plt.xticks(np.arange(len(variance)), labels=[f'{v}%' for v in variance])
plt.legend(loc="lower left", ncols=4)
plt.title("Accuracy average on tesing data")
plt.xlabel("Variance")
plt.ylabel("Accuracy average")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
